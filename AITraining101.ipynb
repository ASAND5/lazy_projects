```python
# 1. Install system dependency 'zstd' and then install Ollama CLI
!sudo apt-get update && sudo apt-get install -y zstd
!curl -fsSL https://ollama.com/install.sh | sh
!pip install ollama

# 2. Start Ollama service in the background
import subprocess
import time

print("Starting Ollama server...")
process = subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
# Give the server a few seconds to initialize
time.sleep(5)

# 3. Pull the required model
print("Pulling model nimble-v3...")
!ollama pull nimble-v3

# 4. Run your code
import ollama

try:
    response = ollama.systemone(
        model='nimble-v3',
        state='Hello World',
        questions={
            'says_hello': {
                'type': 'noul',
                'instructions': 'Does the state text contain a greeting?',
                'criteria': {
                    'true': 'The state text contains a greeting.',
                    'false': 'The state text does not contain a greeting.',
                },
            },
        },
    )
    print("\nResponse:")
    print(response.answers['says_hello'].noul)
except Exception as e:
    print(f"\nAn error occurred: {e}")
```
